```python
import numpy as np
import pandas as pd
```

```python
temperature_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperature_w1)
```

```python
print("Shape:", temperature_w1.shape)
print("Data type:", temperature_w1.dtype)
print("Number of elements:",temperature_w1.size)
```

```python
fahrenheit_w1 = (temperature_w1 * 9/5) + 32
print("fahrenheit:",fahrenheit_w1)
```

```python
print("max:", np.max(temperature_w1))
print("min:", np.min(temperature_w1))
print("mean:", np.mean(temperature_w1))
```

```python
first_three = temperature_w1[0:3]
print("first 3 days:",first_three)
```

```python
weekend = temperature_w1[-2]
print("Weekend:", weekend)
```

```python
middle_3 = temperature_w1[2:5]
print("middle 3 days:",middle_3)
```

```python
marks = pd.Series([95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
             
```

```python

print("1st rank mark(iloc):",marks.iloc[0])
```

```python
top_3 = marks.loc[['Rank1', 'Rank2', 'Rank3']]
print("\ntop 3 rank (iloc):\n", top_3)
```

```python
third_rank = marks.iloc[2]
print("\n3rd rank mark (iloc):", third_rank)
```

```python
above_90 = marks[marks>90]
print("\nmarks>90:\n",above_90)
```

```python
marks.loc['Rank1'] = 100
print("modifying Rank1 to 100:\n", marks)
```

```python
marks= marks.drop('Rank5')
print("\nremoving last rank (Rank5):\n", marks)
```

```python
cgpa = marks / 10
print("\ncgpa (marks / 10):\n", cgpa)
```
